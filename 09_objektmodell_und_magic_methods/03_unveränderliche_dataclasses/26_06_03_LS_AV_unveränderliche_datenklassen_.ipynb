```python
from dataclasses import dataclass


@dataclass
class Name:
    first_name: str
    last_name: str
```

```python
name = Name("John", "Doe")
```

```python
print(name)
```

```python
phone_book = {name: "This is John Doe"}
```

```python
phone_book = {
    name: "This is John Doe"
}
```

```python
print(phone_book)
```

```python
print(phone_book[name])
```

```python
from dataclasses import dataclass


@dataclass(frozen=True)
class Name:
    first_name: str
    last_name: str


name = Name("John", "Doe")

phone_book = {
    name: "This is John Doe"
}
```

```python
print(phone_book)
```

```python
print(phone_book[name])
```
