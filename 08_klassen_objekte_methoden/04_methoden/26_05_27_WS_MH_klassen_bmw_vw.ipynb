```python
"""
Workshop: Kraftfahrzeuge Teil 2
Zusätzlich bekommt die Klasse eine Methode melde_um().
Mit dieser Methode kann man das Kennzeichen eines Fahrzeugs ändern.
"""


class Kfz:
    def __init__(self, hersteller, kennzeichen):
        
        self.hersteller = hersteller
        self.kennzeichen = kennzeichen

    def melde_um(self, neues_kennzeichen):
        """
        Diese Methode ändert das Kennzeichen des Fahrzeugs.

        Das alte Kennzeichen wird überschrieben.
        """

        self.kennzeichen = neues_kennzeichen


def drucke_kfz(kfz):
    """
    Diese Funktion gibt die Daten eines Kraftfahrzeugs aus.
    """

    print(f'Kfz: {kfz.hersteller} mit Kennzeichen "{kfz.kennzeichen}"')



# Hier erzeuge ich drei Fahrzeuge


bmw = Kfz("BMW", "M-BW 123")
bmw2 = Kfz("BMW", "M-BW 123")
vw = Kfz("VW", "WOB-VW 246")


print("Ausgangszustand:")
drucke_kfz(bmw)
drucke_kfz(bmw2)
drucke_kfz(vw)



# Aufgabe 1:
# VW ummelden auf das neue Kennzeichen "BGL-A 9"

print()
print("VW wird umgemeldet:")

vw.melde_um("BGL-A 9")

drucke_kfz(vw)


# Kontrolle:
# Hat das Ummelden wirklich funktioniert?

print()
print("Kontrolle VW:")

print(vw.kennzeichen == "BGL-A 9")


# Aufgabe 2:
# BMW ummelden auf das neue Kennzeichen "F-B 21"


print()
print("BMW wird umgemeldet:")

bmw.melde_um("F-B 21")

drucke_kfz(bmw)
drucke_kfz(bmw2)

# Kontrolle:
# Hat sich bmw2 auch verändert?

print()
print("Kontrolle BMW und BMW2:")

print("Kennzeichen von bmw:", bmw.kennzeichen)
print("Kennzeichen von bmw2:", bmw2.kennzeichen)

print("Sind die Kennzeichen gleich?")
print(bmw.kennzeichen == bmw2.kennzeichen)
```
