```python
class Employee:

    def __init__(self, name, age, gehalt):
        self.name = name
        self.age = age
        self.gehalt = gehalt

    def __str__(self):
        return f"{self.name} ist {self.age} Jahre alt und verdient {self.gehalt} Euro"

    def __repr__(self):
        return f"Employee('{self.name}', {self.age}, {self.gehalt})"

    def __gt__(self, other):
        if not isinstance(other, Employee):
            return NotImplemented

        return self.gehalt > other.gehalt

    def __eq__(self, other):
        if not isinstance(other, Employee):
            return NotImplemented

        return self.name == other.name and self.age == other.age and self.gehalt == other.gehalt

    def gehalt_erhoehen(self, prozent):
        self.gehalt = self.gehalt + self.gehalt * prozent / 100
```

```python
e1 = Employee("Hazemina", 31, 4500)
e2 = Employee("Barbara", 30, 5000)
e3 = Employee("Wael", 31, 5500)
```

```python
print(e1)
print(repr(e1))
```

```python
print(e1 > e2)
print(e1 == e3)
```

```python
e1.gehalt_erhoehen(10)
print(e1)
```
