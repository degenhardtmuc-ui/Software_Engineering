```python
my_string = "Hello, world!"

# str() wandelt etwas in einen normalen String um
print(str(my_string))

# repr() zeigt die technische Darstellung eines Wertes
print(repr(my_string))
```

```python
# str() = schön für Menschen lesbar
# repr() = technische Python-Darstellung(Bei einem String sieht man dann auch die Anführungszeichen)
```
