```python
# Klasse 1: Person: name, age
# default value ist nicht zwingend = 0
class Person:

    def __init__(self, name, age, addresse =None):
        self.name = name
        self.age = age

    def __str__(self):
        return f"{self.name} ist {self.age} Jahre alt"

    def __repr__(self):
        return f"Person('{self.name}', {self.age}, addresse = {self.addresse})"
```

```python
# Klasse 2 Adress: strasse, nr, plz, city 
# str für strasse vermeiden, da es ein keyword in python ist
# Klasse A kennt Klasse B
# Dann hat Klasse A eine Referenz auf B

class Addresse:

    def __init__(self, strasse, nr, plz, city):
        self.str = strasse
        self.nr = nr
        self.plz = plz
        self.city = city

    def __str__(self):
        return f"{self.strasse} {self.nr}, {self.plz} {self.city}"

    def __repr__(self):
        return f"Address('{self.strasse}', {self.nr}, {self.plz}, '{self.city}')"
```

```python
# Klasse 3: Projekt: name

class Projekt:

    def __init__(self, name):
        self.name = name

    def __str__(self):
        return f"Projekt: {self.name}"

    def __repr__(self):
        return f"Projekt('{self.name}')"
```
