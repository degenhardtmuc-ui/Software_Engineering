```python
staedte = [("DE", "München"), ("DE", "Frankfurt"),
    ("FR", "Paris"), ("FR", "Nizza"), ("IT", "Rom")]
```

```python
# Ausgabe: {'DE': ['München', 'Frankfurt'], 'FR': ['Paris', 'Nizza'], 'IT': ['Rom']}
```

```python
def gruppiere(staedte):
    erg = {}

    for land, stadt in staedte:
        if land not in erg:
            erg[land] = []

        erg[land].append(stadt)

    return erg
    
ergebnis = gruppiere(staedte)
```

```python
print(ergebnis)
```
