```python
def pop_element( element, a_list):
    # überprüfe
    if element in a_list:
        element_index = a_list.index(element)
        ergebnis = a_list.pop(element_index)
        
        return ergebnis, element_index
    else:
        print ("Leider das Element", element, "ist nicht in dwer Liste")
```
