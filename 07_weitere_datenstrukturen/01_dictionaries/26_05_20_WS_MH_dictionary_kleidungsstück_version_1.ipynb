```python
def gesamtpreis(kleidungsstücke, preise):
    gesamt = 0

    for teil in kleidungsstücke:
        if teil in preise:
            gesamt = gesamt + preise[teil]

    return gesamt


kleidungsstücke = ["Wintermantel", "Cordhose", "Smoking"]

preise = {"Wintermantel": 80, "Smoking": 40}

ergebnis = gesamtpreis(kleidungsstücke, preise)

print(ergebnis)
```
