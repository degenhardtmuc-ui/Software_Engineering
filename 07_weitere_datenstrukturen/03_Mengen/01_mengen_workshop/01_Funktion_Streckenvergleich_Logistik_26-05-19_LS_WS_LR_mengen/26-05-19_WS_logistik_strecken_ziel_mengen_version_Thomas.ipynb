```python
def roadtrip_switch(triplist: list, goal: int, switch= False):

 """

 Use switch = True for a list of matching neighbours, use False (or skip argument) if you want just the first matching neighbours.

 """

i = 0

matches = []

while i < len(triplist):

if i == len(triplist) - 1:

break

elif triplist[i] + triplist[i+1] == goal and not switch:

return (triplist[i], triplist[i+1])

elif triplist[i] + triplist[i+1] == goal and switch:

matches.append((triplist[i], triplist[i+1]))

i += 1

else:

i += 1

return matches
```
