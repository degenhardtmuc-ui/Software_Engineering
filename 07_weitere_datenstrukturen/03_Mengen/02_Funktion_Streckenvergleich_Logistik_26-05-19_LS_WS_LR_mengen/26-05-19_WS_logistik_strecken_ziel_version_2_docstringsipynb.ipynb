```python
def finde_paare(strecken, ziel):
    """
    Diese Funktion sucht alle möglichen Paare aus einer Liste.

    Die Funktion bekommt:
    - strecken: eine Liste mit Zahlen, zum Beispiel [3, 5, 7, 10]
    - ziel: eine Zahl, die nicht überschritten werden darf

    Die Funktion gibt zurück:
    - eine Liste mit Paaren
    - jedes Paar besteht aus zwei Zahlen
    - beide Zahlen zusammen dürfen höchstens so groß sein wie das Ziel

    Beispiel:
    strecken = [3, 5, 7, 10]
    ziel = 12

    Dann werden Paare gesucht wie:
    (3, 5), (3, 7), (5, 7)

    Aber zum Beispiel nicht:
    (5, 10), weil 5 + 10 = 15 ist und damit größer als 12.
    """

    # Wir erstellen eine leere Liste.
    # In diese Liste kommen später alle gültigen Paare hinein.
    ergebnis = []

    # Die äußere Schleife geht durch alle Positionen der Liste.
    # "links" ist dabei nicht die Zahl selbst, sondern der Index.
    # Der Index ist die Position in der Liste.
    for links in range(len(strecken)):

        # Die innere Schleife startet immer eine Position nach "links".
        # Darum steht hier "links + 1".
        # So vermeiden wir, dass dieselbe Zahl doppelt benutzt wird.
        # Außerdem vermeiden wir doppelte Paare wie (3, 5) und später (5, 3).
        for rechts in range(links + 1, len(strecken)):

            # Hier holen wir die Zahl an der linken Position aus der Liste.
            # Beispiel: Wenn links = 0 ist, dann ist strecken[links] die erste Zahl.
            zahl_links = strecken[links]

            # Hier holen wir die Zahl an der rechten Position aus der Liste.
            # Beispiel: Wenn rechts = 1 ist, dann ist strecken[rechts] die zweite Zahl.
            zahl_rechts = strecken[rechts]

            # Jetzt prüfen wir, ob die beiden Zahlen zusammen kleiner oder gleich dem Ziel sind.
            # Wenn ja, ist dieses Paar erlaubt.
            if zahl_links + zahl_rechts <= ziel:

                # Wenn das Paar erlaubt ist, speichern wir es in der Ergebnisliste.
                # append bedeutet: hinten an die Liste anhängen.
                # Die beiden Zahlen werden als Tupel gespeichert, also zum Beispiel (3, 5).
                ergebnis.append((zahl_links, zahl_rechts))

    # Wenn alle möglichen Paare geprüft wurden,
    # gibt die Funktion die Ergebnisliste zurück.
    return ergebnis
```

```python
strecken = [400, 500, 300, 600, 150, 250]
```

```python
ziel = 650
```

```python
finde_paare(strecken, ziel)
```
