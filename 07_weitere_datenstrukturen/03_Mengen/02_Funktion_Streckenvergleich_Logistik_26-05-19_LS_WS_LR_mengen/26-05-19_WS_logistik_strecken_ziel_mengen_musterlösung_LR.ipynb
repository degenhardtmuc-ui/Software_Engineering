```python
def finde_paare_mit_mengen(strecken, ziel):
    strecken_menge = set(strecken)
    ergebnis = set()

    for erste_zahl in strecken_menge:
        for zweite_zahl in strecken_menge:
            if erste_zahl < zweite_zahl:
                if erste_zahl + zweite_zahl <= ziel:
                    ergebnis.add((erste_zahl, zweite_zahl))

    return ergebnis


strecken = [400, 500, 300, 600, 150, 250]
ziel = 650

paare = finde_paare_mit_mengen(strecken, ziel)

print("Gefundene Paare mit Ziel 650:")

for paar in sorted(paare):
    print(paar, "Summe:", paar[0] + paar[1])
```
