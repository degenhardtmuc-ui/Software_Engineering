```python
def finde_paare_unter_ziel(strecken, ziel):
    passende_paare = []

    for erste_nummer in range(len(strecken)):
        for zweite_nummer in range(erste_nummer + 1, len(strecken)):

            erste_strecke = strecken[erste_nummer]
            zweite_strecke = strecken[zweite_nummer]

            summe = erste_strecke + zweite_strecke

            if summe <= ziel:
                passendes_paar = (erste_strecke, zweite_strecke)
                passende_paare.append(passendes_paar)

    return passende_paare
```
