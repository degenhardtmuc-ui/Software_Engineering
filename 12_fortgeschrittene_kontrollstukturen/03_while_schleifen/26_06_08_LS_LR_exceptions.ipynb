```python
def div():
    
    try: # Happy Path
        zahl = int(input("Gib eine Zahl ein:" ))
        print("Hallo zusammen")
        ergebnis = 100 / zahl
        print("Das Ergebnis ist", ergebnis)
        return ergebnis
    
    except ZeroDivisionError: # None-Happy-Path: wenn was schief geht!
        print("Fehler: Du kannst nicht durch 0 dividieren!")
    
    except ValueError :
        print("Fehler: Du musst eien Zahl angeben!")
```
