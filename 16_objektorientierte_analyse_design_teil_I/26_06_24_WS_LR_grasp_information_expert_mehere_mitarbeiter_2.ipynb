```python
class Mitarbeiter:
    def __init__(self, name: str, stundensatz: float):
        self._name = name
        self._stundensatz = stundensatz

    @property
    def name(self) -> str:
        return self._name

    @property
    def stundensatz(self) -> float:
        return self._stundensatz


class AufwandsBuchung:
    def __init__(self, mitarbeiter: Mitarbeiter, stunden: float):
        self._mitarbeiter = mitarbeiter
        self._stunden = stunden

    def berechne_kosten(self) -> float:
        return self._stunden * self._mitarbeiter.stundensatz

    def zeige_buchung(self):
        print(
            self._mitarbeiter.name,
            "hat",
            self._stunden,
            "Stunden gearbeitet. Kosten:",
            self.berechne_kosten(),
            "Euro"
        )


class Aufgabe:
    def __init__(self, beschreibung: str):
        self._beschreibung = beschreibung
        self._buchungen = []

    def aufwand_buchen(self, mitarbeiter: Mitarbeiter, stunden: float):
        neue_buchung = AufwandsBuchung(mitarbeiter, stunden)
        self._buchungen.append(neue_buchung)

    def berechne_gesamtkosten(self) -> float:
        gesamtkosten = 0

        for buchung in self._buchungen:
            gesamtkosten += buchung.berechne_kosten()

        return gesamtkosten

    def zeige_uebersicht(self):
        print("Aufgabe:", self._beschreibung)
        print("----------------------")

        for buchung in self._buchungen:
            buchung.zeige_buchung()

        print("----------------------")
        print("Gesamtkosten:", self.berechne_gesamtkosten(), "Euro")


# Beispiel zum Ausführen

hiko = Mitarbeiter("Hiko", 120)
anna = Mitarbeiter("Anna", 90)
peter = Mitarbeiter("Peter", 80)

aufgabe1 = Aufgabe("Webseite programmieren")

aufgabe1.aufwand_buchen(hiko, 5)
aufgabe1.aufwand_buchen(anna, 3)
aufgabe1.aufwand_buchen(peter, 4)

aufgabe1.zeige_uebersicht()
```
