```python
# Creator / Erzeuger Pattern:
# A => B : Die Klasse A erstellt Objekte von B wenn
# 1) A besitzt / enthält Objekte von B
# 2) A nutzt intensiv Objekte von B oder sie aufzeichnet (GUI)
# 3) A hat alle notwendigen Daten, um ein Objekt zu erstellen
```

```python
class Profil: # class A

    

    def __init__(self, username: str):

        self.username = username

        self.posts = []

        

    def create_post(self, text): # class B

        post = Post(text) # Creator Pattern

        self.posts.append(post)

        print(f"Post für {self.username} wurde erstellt: '{text}'") 
```

```python
class Post:
    def __init__(self, text:str):
        self.text = text
```

```python
laith_profil = Profil("Laith")
```

```python
laith_profil.create_post("Wir sind kein Schulungsunternehemen mehr! Wir sind KI-Firma!")
```
