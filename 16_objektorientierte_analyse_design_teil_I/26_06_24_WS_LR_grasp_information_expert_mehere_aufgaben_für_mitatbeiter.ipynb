```python
"""
Information Expert Pattern - Mitarbeiter hat mehrere Aufgaben

Neue Idee:
- Eine Aufgabe kann mehrere Mitarbeiter haben.
- Ein Mitarbeiter kann mehrere Aufgaben haben.
- Deshalb gibt es ein Verbindungsobjekt: Arbeitseinsatz.

Ein Arbeitseinsatz bedeutet:
Ein bestimmter Mitarbeiter arbeitet eine bestimmte Anzahl Stunden
an einer bestimmten Aufgabe.
"""


class Mitarbeiter:
    """Ein Mitarbeiter kennt seinen Namen, seinen Stundensatz und seine Einsaetze."""

    def __init__(self, name: str, stundensatz: float):
        self.name = name
        self.stundensatz = stundensatz
        self.einsaetze = []

    def einsatz_hinzufuegen(self, einsatz):
        """Speichert einen Arbeitseinsatz bei diesem Mitarbeiter."""
        self.einsaetze.append(einsatz)

    def berechne_kosten_fuer_stunden(self, stunden: float) -> float:
        """Berechnet, was dieser Mitarbeiter fuer eine bestimmte Stundenzahl kostet."""
        return self.stundensatz * stunden

    def berechne_gesamtstunden(self) -> float:
        """Berechnet alle Stunden, die dieser Mitarbeiter in allen Aufgaben arbeitet."""
        gesamt = 0.0

        for einsatz in self.einsaetze:
            gesamt += einsatz.stunden

        return gesamt

    def berechne_gesamtkosten(self) -> float:
        """Berechnet die Kosten fuer alle Einsaetze dieses Mitarbeiters."""
        gesamt = 0.0

        for einsatz in self.einsaetze:
            gesamt += einsatz.berechne_kosten()

        return gesamt


class Arbeitseinsatz:
    """Verbindet Mitarbeiter, Aufgabe und Stunden miteinander."""

    def __init__(self, mitarbeiter: Mitarbeiter, aufgabe, stunden: float):
        self.mitarbeiter = mitarbeiter
        self.aufgabe = aufgabe
        self.stunden = stunden

    def berechne_kosten(self) -> float:
        """Berechnet die Kosten fuer genau diesen einen Arbeitseinsatz."""
        return self.mitarbeiter.berechne_kosten_fuer_stunden(self.stunden)


class Aufgabe:
    """Eine Aufgabe kennt alle Arbeitseinsaetze, die zu ihr gehoeren."""

    def __init__(self, beschreibung: str):
        self.beschreibung = beschreibung
        self.einsaetze = []

    def mitarbeiter_eintragen(self, mitarbeiter: Mitarbeiter, stunden: float):
        """Traegt einen Mitarbeiter mit Stunden in diese Aufgabe ein."""
        einsatz = Arbeitseinsatz(mitarbeiter, self, stunden)
        self.einsaetze.append(einsatz)
        mitarbeiter.einsatz_hinzufuegen(einsatz)

    def berechne_kosten(self) -> float:
        """Berechnet die Gesamtkosten dieser Aufgabe."""
        gesamt = 0.0

        for einsatz in self.einsaetze:
            gesamt += einsatz.berechne_kosten()

        return gesamt

    def zeige_mitarbeiter(self):
        """Gibt alle Mitarbeiter zurueck, die an dieser Aufgabe arbeiten."""
        namen = []

        for einsatz in self.einsaetze:
            namen.append(einsatz.mitarbeiter.name)

        return namen


class Meilenstein:
    """Ein Meilenstein besteht aus mehreren Aufgaben."""

    def __init__(self, name: str):
        self.name = name
        self.aufgaben = []

    def aufgabe_hinzufuegen(self, aufgabe: Aufgabe):
        """Fuegt dem Meilenstein eine Aufgabe hinzu."""
        self.aufgaben.append(aufgabe)

    def berechne_kosten(self) -> float:
        """Berechnet die Gesamtkosten aller Aufgaben in diesem Meilenstein."""
        gesamt = 0.0

        for aufgabe in self.aufgaben:
            gesamt += aufgabe.berechne_kosten()

        return gesamt


class Projekt:
    """Ein Projekt besteht aus mehreren Meilensteinen."""

    def __init__(self, name: str):
        self.name = name
        self.meilensteine = []

    def meilenstein_hinzufuegen(self, meilenstein: Meilenstein):
        """Fuegt dem Projekt einen Meilenstein hinzu."""
        self.meilensteine.append(meilenstein)

    def berechne_gesamtkosten(self) -> float:
        """Berechnet die Kosten des ganzen Projekts."""
        gesamt = 0.0

        for meilenstein in self.meilensteine:
            gesamt += meilenstein.berechne_kosten()

        return gesamt
```

```python
# Beispiel zum Ausfuehren in VS Code
if __name__ == "__main__":
    # Mitarbeiter erstellen
    heiko = Mitarbeiter("Heiko", 50)
    thomas = Mitarbeiter("Thomas", 80)
    barbara = Mitarbeiter("Barbara", 70)

    # Aufgabe 1: Registrierung
    registrierung = Aufgabe("Registrierung programmieren")
    registrierung.mitarbeiter_eintragen(heiko, 2)
    registrierung.mitarbeiter_eintragen(thomas, 6)

    # Aufgabe 2: Warenkorb
    warenkorb = Aufgabe("Warenkorb programmieren")
    warenkorb.mitarbeiter_eintragen(thomas, 5)
    warenkorb.mitarbeiter_eintragen(barbara, 4)

    # Aufgabe 3: Bezahlfunktion
    bezahlung = Aufgabe("Bezahlfunktion einbauen")
    bezahlung.mitarbeiter_eintragen(barbara, 3)
    bezahlung.mitarbeiter_eintragen(heiko, 2)

    # Aufgabe 4: Testen
    testen = Aufgabe("Programm testen")
    testen.mitarbeiter_eintragen(heiko, 4)
    testen.mitarbeiter_eintragen(thomas, 2)

    # Meilenstein 1: Grundfunktionen
    grundfunktionen = Meilenstein("Grundfunktionen")
    grundfunktionen.aufgabe_hinzufuegen(registrierung)
    grundfunktionen.aufgabe_hinzufuegen(warenkorb)

    # Meilenstein 2: Abschluss
    abschluss = Meilenstein("Abschluss")
    abschluss.aufgabe_hinzufuegen(bezahlung)
    abschluss.aufgabe_hinzufuegen(testen)

    # Projekt erstellen
    projekt = Projekt("Online-Shop")
    projekt.meilenstein_hinzufuegen(grundfunktionen)
    projekt.meilenstein_hinzufuegen(abschluss)

    # Gesamtkosten des Projekts ausgeben
    print("Projekt:", projekt.name)
    print("Gesamtkosten:", projekt.berechne_gesamtkosten())

    # Zusaetzlich: Auswertung pro Mitarbeiter
    print("\nAuswertung pro Mitarbeiter:")

    for mitarbeiter in [heiko, thomas, barbara]:
        print(mitarbeiter.name)
        print("  Stunden:", mitarbeiter.berechne_gesamtstunden())
        print("  Kosten:", mitarbeiter.berechne_gesamtkosten())
```
