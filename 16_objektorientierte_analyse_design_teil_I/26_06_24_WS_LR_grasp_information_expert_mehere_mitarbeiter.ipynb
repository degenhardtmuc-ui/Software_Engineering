```python
# Aufgabe:  warum ist dieser Code hier nicht gut?

# Verbessert den Code so, dass die "Information Expert" Pattern Anwendung findet!

 

# --- REINE DATENKLASSEN (Kein Information Expert) ---

class Mitarbeiter:

    def __init__(self, name: str, stundensatz: float):

        self.name = name

        self.stundensatz = stundensatz

 

class Aufgabe:

    def __init__(self, beschreibung: str, mitarbeiter: Mitarbeiter, stunden: float):

        self.beschreibung = beschreibung

        self.mitarbeiter = mitarbeiter

        self.stunden = stunden
```

```python
class Meilenstein:

    def __init__(self, name: str):

        self.name = name

        self.aufgaben = []

 

class Projekt:

    def __init__(self, name: str):

        self.name = name

        self.meilensteine = []

 

# --- DIE KATASTROPHE ---

class ProjektController:

    def berechne_projekt_gesamtkosten(self, projekt: Projekt) -> float:

        gesamtkosten = 0.0

        # Der Controller muss die komplette Struktur der Anwendung kennen!

        for meilenstein in projekt.meilensteine:

            for aufgabe in meilenstein.aufgaben:

                # Tiefes Hineingreifen in fremde Objekte (Verletzung von Law of Demeter)

                stundensatz = aufgabe.mitarbeiter.stundensatz

                gesamtkosten += aufgabe.stunden * stundensatz

        return gesamtkosten
```

```python
# ======== Information Expert============

# Projektkosten mit mehreren Mitarbeitern - Information Expert Pattern
# - Jeder Mitarbeiter kennt seinen Stundensatz
# - Jeder Arbeitseinsatz kennt Mitarbeiter + Stunden
# - Jede Aufgabe kennt ihre Arbeitseinsätze
# - Jeder Meilenstein kennt seine Aufgaben
# - Das Projekt kennt seine Meilensteine
# Dadurch berechnet immer die Klasse etwas, die die passenden Informationen besitzt
```

```python
# Klasse Mitarbeiter
```

```python
class Mitarbeiter:
   

    def __init__(self, name: str, stundensatz: float):
        self.name = name
        self.stundensatz = stundensatz

    def berechne_kosten_für_stunden(self, stunden: float) -> float:
        
        return self.stundensatz * stunden
```

```python
# Klasse Arbeitseinsatz
```

```python
class Arbeitseinsatz:
    

    def __init__(self, mitarbeiter: Mitarbeiter, stunden: float):
        self.mitarbeiter = mitarbeiter
        self.stunden = stunden

    def berechne_kosten(self) -> float:
        
        return self.mitarbeiter.berechne_kosten_für_stunden(self.stunden)
```

```python
# Klasse Aufgabe
```

```python
class Aufgabe:
    

    def __init__(self, beschreibung: str):
        self.beschreibung = beschreibung
        self.einsätze = []

    def mitarbeiter_eintragen(self, mitarbeiter: Mitarbeiter, stunden: float):
        
        einsatz = Arbeitseinsatz(mitarbeiter, stunden)
        self.einsätze.append(einsatz)

    def berechne_kosten(self) -> float:
        
        gesamt = 0.0

        for einsatz in self.einsaetze:
            gesamt += einsatz.berechne_kosten()

        return gesamt
```

```python
# Klasse Meilenstein
```

```python
class Meilenstein:
    

    def __init__(self, name: str):
        self.name = name
        self.aufgaben = []

    def aufgabe_hinzufügen(self, aufgabe: Aufgabe):
        
        self.aufgaben.append(aufgabe)

    def berechne_kosten(self) -> float:
        
        gesamt = 0.0

        for aufgabe in self.aufgaben:
            gesamt += aufgabe.berechne_kosten()

        return gesamt
```

```python
class Projekt:
    
    def __init__(self, name: str):
        self.name = name
        self.meilensteine = []

    def meilenstein_hinzufügen(self, meilenstein: Meilenstein):
        
        self.meilensteine.append(meilenstein)

    def berechne_gesamtkosten(self) -> float:
        
        gesamt = 0.0

        for meilenstein in self.meilensteine:
            gesamt += meilenstein.berechne_kosten()

        return gesamt
```
