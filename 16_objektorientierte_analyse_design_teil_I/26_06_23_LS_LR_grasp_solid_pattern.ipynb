```python
# ==========================================
# GRASP + SOLID Beispiel: Artikel und Warenkorb
# ==========================================


class Artikel:
    """
    Die Klasse Artikel beschreibt einen einzelnen Artikel.
    Ein Artikel hat:
    - einen Namen
    - einen Preis
    - eine Anzahl
    """

    def __init__(self, name, preis, anzahl):
        self.name = name
        self.preis = preis
        self.anzahl = anzahl

    def berechne_zwischensumme(self):
        return self.preis * self.anzahl

    def zeige_artikel(self):
        print(f"{self.anzahl} x {self.name} für {self.preis} € = {self.berechne_zwischensumme()} €")
```

```python
class Warenkorb:
    """
    Die Klasse Warenkorb sammelt mehrere Artikel.
    Der Warenkorb kann:
    - Artikel hinzufügen
    - den Gesamtpreis berechnen
    - alle Artikel anzeigen
    """

    def __init__(self):
        self.artikel_liste = []

    def add_artikel(self, artikel):
        self.artikel_liste.append(artikel)

    def berechne_gesamtpreis(self):
        gesamtpreis = 0

        for artikel in self.artikel_liste:
            gesamtpreis = gesamtpreis + artikel.berechne_zwischensumme()

        return gesamtpreis

    def zeige_warenkorb(self):
        print("----- Warenkorb -----")

        for artikel in self.artikel_liste:
            artikel.zeige_artikel()

        print("---------------------")
        print(f"Gesamtpreis: {self.berechne_gesamtpreis()} €")
```

```python
# ==========================================
# Hauptprogramm
# ==========================================

artikel_1 = Artikel("Apfel", 1.50, 4)
artikel_2 = Artikel("Brot", 3.20, 2)
artikel_3 = Artikel("Milch", 1.10, 3)

warenkorb = Warenkorb()

warenkorb.add_artikel(artikel_1)
warenkorb.add_artikel(artikel_2)
warenkorb.add_artikel(artikel_3)

warenkorb.zeige_warenkorb()
```
