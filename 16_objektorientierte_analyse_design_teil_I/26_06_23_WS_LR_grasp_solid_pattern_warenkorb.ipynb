```python
class Artikel:
    def __init__(self, name: str, preis: float):
        self.name = name
        self.preis = preis


class Warenkorb:
    def __init__(self):
        self.artikel_liste = []

    def fuege_artikel_hinzu(self, artikel: Artikel):
        self.artikel_liste.append(artikel)

    def berechne_total(self) -> float:
        total = 0.0

        for artikel in self.artikel_liste:
            total += artikel.preis

        return total
```

```python
apfel = Artikel("Apfel", 1.50)
brot = Artikel("Brot", 3.00)
```

```python
warenkorb = Warenkorb()
```

```python
warenkorb.fuege_artikel_hinzu(apfel)
warenkorb.fuege_artikel_hinzu(brot)
```

```python
print(warenkorb.berechne_total())
```
