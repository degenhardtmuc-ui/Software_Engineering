```python
anzahl_piraten = 7
anzahl_kapitän = 1
anzahl_personen = anzahl_piraten + anzahl_kapitän
```

```python
schatz = 1000
```

```python
anteil_pro_person = schatz // anzahl_personen
```

```python
rest_fuer_kapitän = schatz % anzahl_personen
```

```python
kapitän_bekommt = anteil_pro_person + rest_fuer_kapitän
```

```python
print("Ausgangssituation:")
print("Personen insgesamt:", anzahl_personen)
print("Jeder Pirat bekommt:", anteil_pro_person, "Golddublonen")
print("Der Kapitän bekommt extra:", rest_fuer_kapitaen, "Golddublonen")
print("Der Kapitän bekommt insgesamt:", kapitän_bekommt, "Golddublonen")

print()
```
