```python
# 09 Zusicherungen / Assertions
# Mini-Workshop

# Gegeben:
my_int = 1
my_float = 1.0

# Eine Assertion prüft, ob eine Bedingung wahr ist.
# Wenn die Bedingung True ist, passiert nichts.
# Wenn die Bedingung False ist, kommt ein AssertionError.

# 1. Eigenschaft:
# my_int == 1
# 1 ist gleich 1, also ist die Aussage wahr.
assert my_int == 1

# 2. Eigenschaft:
# my_float == my_int
# 1.0 und 1 sind vom Wert her gleich.
# Python sagt deshalb: True
assert my_float == my_int

# 3. Eigenschaft:
# my_float == "1.0"
# Diese Aussage ist falsch.
# my_float ist eine Zahl: 1.0
# "1.0" ist ein Text/String.
# Deshalb sichern wir die Negation zu:
assert my_float != "1.0"

print("Alle Assertions wurden erfolgreich geprüft.")
print("Es wurde kein Fehler ausgelöst.")
```
