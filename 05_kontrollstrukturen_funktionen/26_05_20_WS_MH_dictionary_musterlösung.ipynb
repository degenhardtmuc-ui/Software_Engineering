```python
def gesamtpreis(kleidungsstücke: list, preise: dict) -> float:
    """
    Berechnet den Gesamtpreis aller Kleidungsstücke,
    die in der Preisliste vorhanden sind.

    kleidungsstücke: Liste mit Kleidungsstücken, z. B. ["Wintermantel", "Cordhose", "Smoking"]
    preise: Dictionary mit Preisen, z. B. {"Wintermantel": 80, "Smoking": 40}

    Rückgabe:
    Gesamtpreis als Zahl.
    """

    gesamt = 0

    for kleidungsstück in kleidungsstücke:
        preis = preise.get(kleidungsstück, 0)
        gesamt = gesamt + preis

    return gesamt


# Beispiel aus der Aufgabe
kleidungsstücke = ["Wintermantel", "Cordhose", "Smoking"]

preise = {
    "Wintermantel": 80,
    "Smoking": 40
}

ergebnis = gesamtpreis(kleidungsstücke, preise)

print("Der Gesamtpreis beträgt:", ergebnis)
```
