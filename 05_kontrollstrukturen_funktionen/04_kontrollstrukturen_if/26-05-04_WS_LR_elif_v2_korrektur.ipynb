```python
def Beitragsberechner_Steuer():
    gehalt = int(input("Wie hoch ist dein Gehalt?"))
    if gehalt > 60000:
        print("Mitgliedschaft ist 300€")
    elif gehalt >= 50000 and gehalt <= 60000:
        print("Mitgliedschaft ist 250€")
    elif gehalt >= 40000 and gehalt < 50000:
        print("Mitgliedschaft ist 200€")
    elif gehalt >= 30000 and gehalt < 40000:
        print("Mitgliedschaft ist 150€")
    elif gehalt >= 20000 and gehalt < 30000:
        print("Mitgliedschaft ist 100€")
    elif gehalt >= 15000 and gehalt < 20000:
        print("Mitgliedschaft ist 50€")
    else:
        print("Für Sie ist das nun kostenfrei!")
```

```python
Beitragsberechner_Steuer()
```
