```python
def effektive_spende(n): 
    # ZRD doubles the donation: n * 2
    # YB3 adds 10 euros: + 10
    # Total effective donation: n * 2 + 10 
    return n * 2 + 10
```

```python
print(effektive_spende(20))
```

```python
print(effektive_spende(10))
```

```python
print(effektive_spende(25))
```
