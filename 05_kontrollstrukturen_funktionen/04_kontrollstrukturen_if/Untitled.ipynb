```python
def fits_in_line(text: str, line_length: int = 72):
    
    if len(text) <= line_length:
        return True
    else:
        return False



print(fits_in_line("Hallo"))       # Erwartet: True, weil 5 <= 72
print(fits_in_line("Hallo", 3))    # Erwartet: False, weil 5 > 3
print(fits_in_line("abcd", 4))     # Erwartet: True, weil 4 <= 4
print(fits_in_line("abcd", 2))     # Erwartet: False, weil 4 > 2
```
