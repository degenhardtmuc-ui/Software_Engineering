```python
def addiere(a, b):
    return a + b
    
print(addiere(5, 3))
```

```python
def subtrahiere(a, b):
    return a - b
    
print(subtrahiere(5, 3))
```

```python
def multipliziere(a, b):
    return a * b

print(multipliziere(5, 3))
```

```python
def dividiere(a, b):
    if b != 0:
        return a / b
    else:
        return "Fehler: Division durch 0"
        
print(dividiere(5, 3))
```

```python
def rest(a, b):
    if b != 0:
        return a % b
    else:
        return "Fehler: Division durch 0"
print(rest(5, 3)) 
```
