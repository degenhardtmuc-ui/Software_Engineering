```python
import argparse


def question_1():
    """
    Question: Bestimme den Betrag von -3.5.
    """
    print(f"Der Betrag von -3.5 ist: {abs(-3.5)}")
    return


def question_2():
    """
    Question: Bestimme das Maximum der Werte 2.5 und 1.7.
    """
    print("Um das Maximum der Werte 2.5 und 1.7 zu bestimmen, kann man max() nutzen.")
    print(f"Mit max(2.5, 1.7) erhält man also: {max(2.5, 1.7)}")
    return


def question_3():
    """
    Question: Wandle den String '1.234' in eine Zahl um und addiere 2.345.
    """
    value = "1.234"
    zahl = float(value)
    ergebnis = zahl + 2.345

    print(f"Der String '{value}' wird mit float() zur Zahl: {zahl}")
    print(f"Die Addition ergibt: {ergebnis}")
    return


def bonus(value):
    """
    Bonus: Runde .5 immer zur nächsten höheren ganzen Zahl.
    Beispiel: 2.5 -> 3, 3.5 -> 4, 4.4 -> 4
    """
    if value % 1 == 0.5:
        ergebnis = int(value) + 1
    else:
        ergebnis = round(value)

    print(f"Der Wert {value} gerundet ist: {ergebnis}")
    return ergebnis


def start():
    question_1()
    print()

    question_2()
    print()

    question_3()
    print()

    bonus(2.5)
    bonus(3.5)
    bonus(4.4)


if __name__ == "__main__":
    start()
```
