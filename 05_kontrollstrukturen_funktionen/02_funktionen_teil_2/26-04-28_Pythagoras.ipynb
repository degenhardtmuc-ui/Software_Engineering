```python
def berechne_pythagoras (kathete1, kathete2):
    hypotenuse = ( (kathete1 **2) + (kathete2))**0.5
    return hypotenuse
    
```

```python
def berechne_zaunlänge(str1, str2):
    str3 = berechne_pythagoras(str1, str2)
    zaun = str1 + str2 + str3
    return zaun
```

```python
land_str = 30
```

```python
adenauer_str = 40
```

```python
berechne_zaunlänge(land_str, adenauer_str)
```
