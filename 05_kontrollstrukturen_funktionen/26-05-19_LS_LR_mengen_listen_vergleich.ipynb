```python
# listen Vergleich
```

```python
def vergleich(lotto, spieler):

    ergebnis = []

    for item in lotto:

        if item in spieler:

            ergebnis.append(item)

    return len(ergebnis), ergebnis
```

```python
lotto = [5, 17, 33, 45, 47, 49]
```

```python
spieler = [13, 33, 40, 45, 48, 49]
```

```python
print(vergleich(lotto, spieler))
```

```python
def vergleich(lotto, spieler):
    ergebnis = set(lotto) & set(spieler)
    return len(ergebnis), ergebnis
```

```python
spieler = [13, 33, 40, 45, 48, 49]
```

```python
lotto = [5, 17, 33, 45, 47, 49]
```

```python
ergebnis = vergleich(lotto, spieler)
```

```python
print(vergleich(lotto, spieler))
```
