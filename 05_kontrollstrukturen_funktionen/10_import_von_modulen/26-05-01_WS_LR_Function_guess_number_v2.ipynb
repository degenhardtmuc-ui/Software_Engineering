```python
import random
```

```python
def get_computer_nr():                                  # Diese Funktion erzeugt die Zahl vom Computer.
    return random.randint(1, 10)
```

```python
def bewerte(usr_nr, computer_nr):
    if usr_nr == computer_nr:
        print("Richtig geraten!")
        return True
    elif usr_nr < computer_nr:
        print("Zu niedrig!")
        return False
    else:
        print("Zu hoch!")
        return False
```

```python
def start_v2():
    computer_nr = get_computer_nr()

    for versuch in range(1, 4):
        print(f"Versuch {versuch} von 3")

        usr_nr = get_user_nr()

        if bewerte(usr_nr, computer_nr):
            print("Du hast gewonnen!")
            break
    else:
        print(f"Leider verloren. Die richtige Zahl war {computer_nr}.")
```
