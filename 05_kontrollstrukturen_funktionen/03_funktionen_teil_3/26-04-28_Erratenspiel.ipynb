```python
import random
```

```python
computer_zahl = random.randint(1, 10)
```

```python
user_zahl = int(input("Gib mir eine Zahl zwischen 1 und 10:"))
```

```python
if user_zahl == computer_zahl:
    print("Yes! Du hast die Zahl korrekt eingegeben.")
else:
    print("Oh no! Du hast leider eine falsche Zahl eingegeben.")
    print("Die korrekte Zahl lautet", computer_zahl)
```
