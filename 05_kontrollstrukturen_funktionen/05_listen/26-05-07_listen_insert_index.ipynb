```python
# def insert_next_to(liste, element) : insert_text_to(d, "Uwe"): fügt Uwe rechts von der ersten Uwe
# sollte "Uwe" nicht in der Liste sein, dann wird Uwe an einer random Postion gespeichert!

# Aufgabe: Element neben einem gleichen Element einfügen
# Beispiel: insert_next_to(d, "Uwe")
# soll "Uwe" zwei Positionen RECHTS von der ersten "Uwe"-Stelle einfügen.
# Wenn "Uwe" NICHT in der Liste ist:
# Dann wird "Uwe" an einer zufälligen Position eingefügt.

import random


def insert_next_to(liste, element):
   
    if element in liste:
        erste_position = liste.index(element)
        neue_position = erste_position + 2

        liste.insert(neue_position, element)

    else:
       zufalls_position = random.randint(0, len(liste))

        
        liste.insert(zufalls_position, element)
```

```python
d = ["Barbara", "Hazemina", "Tanju", "Wael", "Uwe", "Thomas", "Uwe", "Heiko", "Leonid", "Daniel", "Uwe"]
```

```python
insert_next_to(d, "Uwe")
```

```python
# Aufgabe: Element direkt neben dem ersten gleichen Element einfügen
# Diese Version benutzt index().

import random


def insert_next_to_index(liste, element):
    # 1. Prüfen: Gibt es das Element schon in der Liste?
    if element in liste:
        # 2. index() findet die ERSTE Position vom Element.
        position = liste.index(element)

        # 3. + 1 bedeutet:
        #    eine Stelle rechts neben dem gefundenen Element.
        neue_position = position + 1

        # 4. insert() fügt das Element an dieser Stelle ein.
        liste.insert(neue_position, element)

    else:
        # 5. Wenn das Element NICHT in der Liste ist:
        #    Dann wählen wir eine zufällige Position.
        zufalls_position = random.randint(0, len(liste))

        # 6. Dort wird das Element eingefügt.
        liste.insert(zufalls_position, element)


d = ["Barbara", "Hazemina", "Tanju", "Wael", "Uwe", "Thomas", "Uwe", "Heiko", "Leonid", "Daniel", "Uwe"]

insert_next_to_index(d, "Uwe")

print(d)
```
