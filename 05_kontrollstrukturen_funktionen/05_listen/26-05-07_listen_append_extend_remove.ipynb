```python
def entferne_alle(liste, element):
    wurde_entfernt = False
    while element in liste:
        liste.remove(element)
        wurde_entfernt = True
    return wurde_entfernt
```

```python
d = [
    "Barbara",
    "Hazemina",
    "Tanju",
    "Wael",
    "Thomas",
    "Heiko",
    "Leonid",
    "Daniel",
    "Uwe",
    "Uwe",
    "Uwe"]
```

```python
print("Liste vorher:")
print(d)
```

```python
ergebnis = entferne_alle(d, "Uwe")

print("Liste nachher:")
print(d)
```

```python
if ergebnis == True:
    print("Das Element wurde entfernt.")
else:
    print("Das Element war nicht in der Liste.")
```
