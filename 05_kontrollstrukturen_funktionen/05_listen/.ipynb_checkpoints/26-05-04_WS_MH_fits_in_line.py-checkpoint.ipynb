```python
# Workshop: fits_in_line


def fits_in_line(text: str, line_length: int = 72):
 
    if len(text) <= line_length:
        return True
    else:
        return False



print(fits_in_line("Hallo"))       
print(fits_in_line("Hallo", 3))    
print(fits_in_line("abcd", 4))     
print(fits_in_line("abcd", 2))     
```
