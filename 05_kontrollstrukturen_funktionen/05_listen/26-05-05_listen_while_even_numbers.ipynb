```python
def save_even_numbers_to_list(n):
    even_numbers = []  # Leere Liste für die geraden Zahlen

    while n > 0:  # Solange n größer als 0 ist
        if n % 2 == 0:  # Prüfen: Ist n gerade?
            even_numbers.append(n)  # Gerade Zahl in die Liste einfügen

        n = n - 1  # n wird um 1 kleiner gemacht

    return even_numbers  # Fertige Liste zurückgeben


print(save_even_numbers_to_list(10))
```
