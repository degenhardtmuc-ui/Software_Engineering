```python
class DeckungsError(Exception):

    

    def __init__(self, konto_stand, betrag, msg="Sie haben nicht genung Guthaben!"):

        self.konto_stand = konto_stand

        self.betrag = betrag

        self.msg = msg

        super().__init__(self.msg)

        

    def __str__(self):

        return f"{self.msg}(Aktuelles Guthaben: {self.konto_stand}€, Gewünschter Betrag: {self.betrag}€)"
```

```python
class Konto:

    

    def __init__(self, iban, konto_stand):

        self.iban = iban

        self.konto_stand = konto_stand

        

    def geld_abheben(self, betrag):

        if self.konto_stand < betrag:

            # Best Practice: werfe ein Exception: präzise eigene!

            raise DeckungsError(self.konto_stand, betrag)

        print("Abheben ist erfolgreich! Du hast genügend Deckung")

        self.konto_stand = self.konto_stand - betrag

        print("Dein neuer Konto-Stand beträgt:", self.konto_stand)

        

    def geld_einzahlen(self, betrag):

        self.konto_stand = self.konto_stand + betrag
```

```python
meine_konto = Konto ("DE12345", 1_000)
```

meine_konto.geld_abheben(800)

```python
try:

    meine_konto = Konto("DE24 345 678 120", 1_000)

    meine_konto.geld_abheben(2_000)

except DeckungsError as error:

    print(f"Fehler abgefangen: {error}")

    meine_konto.geld_einzahlen(1_500)

    meine_konto.geld_abheben(2_000)





```
