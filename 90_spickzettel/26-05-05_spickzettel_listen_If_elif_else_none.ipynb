```python
# Alles kombiniert

def check_zahl(zahl):
    if zahl > 10:
        return "groß"
    elif zahl == 10:
        return "genau 10"
    else:
        return None
```

```python
print(check_zahl(15))
```

```python
print(check_zahl(10))
```

```python
print(check_zahl(5))
```
