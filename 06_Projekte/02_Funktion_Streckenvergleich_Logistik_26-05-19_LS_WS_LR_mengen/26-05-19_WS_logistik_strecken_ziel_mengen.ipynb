```python
def finde_paare_mit_mengen(strecken, ziel):
    """
    Diese Funktion sucht alle möglichen Paare aus einer Liste von Strecken.

    Die Funktion arbeitet mit Mengen.

    Das bedeutet:
    - Aus der Liste wird zuerst eine Menge gemacht.
    - Doppelte Zahlen würden automatisch entfernt werden.
    - Danach werden alle möglichen Zahlenpaare geprüft.

    Ein Paar ist gültig, wenn:
    - die erste Zahl kleiner ist als die zweite Zahl
    - beide Zahlen zusammen kleiner oder gleich dem Ziel sind

    Beispiel:
    strecken = [400, 500, 300, 600, 150, 250]
    ziel = 650

    Dann ist zum Beispiel erlaubt:
    150 + 500 = 650

    Aber nicht erlaubt:
    400 + 500 = 900
    """

    # Aus der Liste "strecken" machen wir eine Menge.
    # Eine Menge speichert jeden Wert nur einmal.
    strecken_menge = set(strecken)

    # Hier erstellen wir eine leere Menge für die Ergebnisse.
    # In diese Menge kommen später alle gültigen Paare hinein.
    ergebnis = set()

    # Die erste Schleife nimmt eine Zahl aus der Menge.
    # Diese Zahl nennen wir "erste_zahl".
    for erste_zahl in strecken_menge:

        # Die zweite Schleife nimmt ebenfalls eine Zahl aus der Menge.
        # Diese Zahl nennen wir "zweite_zahl".
        for zweite_zahl in strecken_menge:

            # Wir prüfen, ob die erste Zahl kleiner ist als die zweite Zahl.
            #
            # Warum?
            # Damit keine doppelten Paare entstehen.
            #
            # Beispiel:
            # (150, 400) soll gespeichert werden.
            # (400, 150) soll NICHT nochmal gespeichert werden.
            #
            # Außerdem verhindert diese Bedingung Paare wie:
            # (150, 150)
            if erste_zahl < zweite_zahl:

                # Jetzt prüfen wir, ob beide Zahlen zusammen
                # kleiner oder gleich dem Ziel sind.
                #
                # Beispiel:
                # 150 + 500 = 650
                #
                # Das ist erlaubt, weil 650 <= 650 ist.
                if erste_zahl + zweite_zahl <= ziel:

                    # Wenn das Paar erlaubt ist,
                    # speichern wir es in der Ergebnis-Menge.
                    #
                    # Bei Mengen benutzt man add().
                    # Bei Listen würde man append() benutzen.
                    ergebnis.add((erste_zahl, zweite_zahl))

    # Am Ende gibt die Funktion alle gefundenen Paare zurück.
    return ergebnis


# ---------------------------------------------------------
# Hauptprogramm
# ---------------------------------------------------------

# Das sind die vorgegebenen Strecken aus der Aufgabe.
strecken = [400, 500, 300, 600, 150, 250]

# Das ist das Ziel.
# Zwei Strecken zusammen dürfen höchstens 650 ergeben.
ziel = 650

# Hier rufen wir die Funktion auf.
# Das Ergebnis speichern wir in der Variable "paare".
paare = finde_paare_mit_mengen(strecken, ziel)

# Hier geben wir eine Überschrift aus.
print("Gefundene Paare mit Ziel 650:")

# Eine Menge hat keine feste Reihenfolge.
# Darum benutzen wir sorted(), damit die Ausgabe ordentlich sortiert ist.
for paar in sorted(paare):

    # paar[0] ist die erste Zahl im Paar.
    # paar[1] ist die zweite Zahl im Paar.
    print(paar, "Summe:", paar[0] + paar[1])
```
