```python
def finde_strecken_paar(strecken, ziel):

    fahren = set()

    ergebnis = []

    for zahl in strecken:

        moeglich = ziel - zahl

        if moeglich in fahren:

            ergebnis.append((moeglich, zahl))

        fahren.add(zahl)

    return ergebnis if ergebnis else None
```
