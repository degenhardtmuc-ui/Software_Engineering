```python
entfernungen = [300, 500, 600, 270, 800, 150]


def finde_strecken_paar(strecken, ziel):
    # Wir erstellen eine leere Menge.
    # Hier speichern wir Strecken, die wir schon markiert haben.
    markierte_strecken = set()

    # Wir gehen jede Strecke aus der Liste einzeln durch.
    for strecke in strecken:

        # Wir markieren die aktuelle Strecke,
        # indem wir sie in die Menge legen.
        markierte_strecken.add(strecke)

        # Wir berechnen, welche Strecke noch fehlt,
        # damit wir zusammen auf das Ziel kommen.
        fehlende_strecke = ziel - strecke

        # Wir prüfen:
        # Ist diese fehlende Strecke schon in der Menge?
        if fehlende_strecke in markierte_strecken:

            # Wenn ja, haben wir ein passendes Paar gefunden.
            return (strecke, fehlende_strecke)

    # Wenn kein Paar gefunden wurde, geben wir None zurück.
    return None


print(finde_strecken_paar(entfernungen, 600))
```
