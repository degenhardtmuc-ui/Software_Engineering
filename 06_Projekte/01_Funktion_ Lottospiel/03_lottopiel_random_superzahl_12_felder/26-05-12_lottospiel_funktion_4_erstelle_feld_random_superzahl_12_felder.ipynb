```python
def erstelle_feld_random():
    lottozahlen = []

    while len(lottozahlen) < 6:
        zahl = random.randint(1, 49)

        if zahl not in lottozahlen:
            lottozahlen.append(zahl)

    superzahl = random.randint(1, 9)

    lottozahlen.sort()

    feld = [lottozahlen, superzahl]

    return feld
```
