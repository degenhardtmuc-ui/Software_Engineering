```python
# 1.) Spieler gibt 6 Zahlen ein - mit Beruecksichtigung, ob diese doppelt sind oder keine Zahlen zwischen 1 und 49 sind.

 

 

def spieler_zahlen():

    zahlen = []

 

    print("Lieber Spieler, gib bitte 6 verschiedene Zahlen zwischen 1 und 49 ein:")

 

    while len(zahlen) < 6:

        zahl = int(input(f"Zahl {len(zahlen)+1}: "))

 

        if zahl < 1 or zahl > 49:

            print("Die Zahl muss zwischen 1 und 49 liegen.")

        elif zahl in zahlen:

            print("Die Zahl wurde bereits eingegeben.")

        else:

            zahlen.append(zahl)

 

    return zahlen
```

```python
# 2.) Computer generiert 6 Zufallszahlen - ohne verwenden von  sample

 

# Computer/PC ist generator

 

import random

 

def generator_zahlen():

    zahlen = []

 

    for _ in range(6):

        zahl = random.randint(1, 49)

        zahlen.append(zahl)

 

    return zahlen

 

 

# Beispiel

 

print(generator_zahlen())
```
