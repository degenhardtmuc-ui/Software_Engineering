```python
# Aufgabe: Vererbung mit Mitarbeiter, Arbeiter und Manager


class Mitarbeiter:
    def __init__(self, name, personalnummer, grundgehalt):
        self.name = name
        self.personalnummer = personalnummer
        self.grundgehalt = grundgehalt

    def salary(self):
        return self.grundgehalt * 13 / 12

    def __str__(self):
        return f"{self.name} hat ein Gehalt von {self.salary():.2f} Euro."

    def __repr__(self):
        return f"Mitarbeiter({self.name}, {self.personalnummer}, {self.grundgehalt})"


class Arbeiter(Mitarbeiter):
    def __init__(self, name, personalnummer, grundgehalt, ueberstunden, stundenlohn):
        super().__init__(name, personalnummer, grundgehalt)
        self.ueberstunden = ueberstunden
        self.stundenlohn = stundenlohn

    def salary(self):
        return super().salary() + self.ueberstunden * self.stundenlohn

    def __repr__(self):
        return f"Arbeiter({self.name}, {self.personalnummer}, {self.grundgehalt}, {self.ueberstunden}, {self.stundenlohn})"


class Manager(Mitarbeiter):
    def __init__(self, name, personalnummer, grundgehalt, bonus):
        super().__init__(name, personalnummer, grundgehalt)
        self.bonus = bonus

    def salary(self):
        return super().salary() + self.bonus

    def __repr__(self):
        return f"Manager({self.name}, {self.personalnummer}, {self.grundgehalt}, {self.bonus})"


# Arbeiter erzeugen
hans = Arbeiter("Hans", 123, 36000.0, 3.5, 40.0)

print(hans)
print("Gehalt von Hans:", hans.salary())


# Assertions für Arbeiter
assert hans.name == "Hans"
assert hans.personalnummer == 123
assert hans.grundgehalt == 36000.0
assert hans.ueberstunden == 3.5
assert hans.stundenlohn == 40.0
assert hans.salary() == 39140.0


# Manager erzeugen
sepp = Manager("Sepp", 843, 60000.0, 30000.0)

print(sepp)
print("Gehalt von Sepp:", sepp.salary())


# Assertions für Manager
assert sepp.name == "Sepp"
assert sepp.personalnummer == 843
assert sepp.grundgehalt == 60000.0
assert sepp.bonus == 30000.0
assert sepp.salary() == 95000.0
```
