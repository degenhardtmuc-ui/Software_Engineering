```python
class Adresse:
    def __init__(self, strasse, plz, ort):
        self.strasse = strasse
        self.plz = plz
        self.ort = ort

    def __repr__(self):
        return f"Adresse('{self.strasse}', '{self.plz}', '{self.ort}')"
```

```python
class Projekt:
    def __init__(self, name, beschreibung):
        self.name = name
        self.beschreibung = beschreibung

    def __repr__(self):
        return f"Projekt('{self.name}', '{self.beschreibung}')"
```

```python
class Mitarbeiter:
    def __init__(self, vorname, nachname, personalnummer, projekt=None):
        self.vorname = vorname
        self.nachname = nachname
        self.personalnummer = personalnummer
        self.adressen = []
        self.projekt = projekt

    def add_adresse(self, adresse):
        if isinstance(adresse, Adresse):
            self.adressen.append(adresse)

    def remove_adresse(self, adresse):
        if adresse in self.adressen:
            self.adressen.remove(adresse)

    def set_projekt(self, projekt):
        if isinstance(projekt, Projekt):
            self.projekt = projekt

    def __repr__(self):
        return f"Mitarbeiter('{self.vorname}', '{self.nachname}', {self.personalnummer})"
```

```python
class Firma:
    def __init__(self, name, adresse=None, bekannter_mitarbeiter=None):
        self.name = name
        self.adresse = adresse
        self.mitarbeiter = []

    def add_mitarbeiter(self, mitarbeiter):
        if isinstance(mitarbeiter, Mitarbeiter):
            self.mitarbeiter.append(mitarbeiter)

    def remove_mitarbeiter(self, mitarbeiter):
        if mitarbeiter in self.mitarbeiter:
            self.mitarbeiter.remove(mitarbeiter)

    def set_adresse(self, adresse):
        if isinstance(adresse, Adresse):
            self.adresse = adresse

    def __repr__(self):
        return f"Firma('{self.name}')"
```
