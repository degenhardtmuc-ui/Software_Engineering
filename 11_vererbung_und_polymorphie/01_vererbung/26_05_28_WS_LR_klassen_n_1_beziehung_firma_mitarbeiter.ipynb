```python
class Adresse:
    def __init__(self, strasse, plz, ort):
        self.strasse = strasse
        self.plz = plz
        self.ort = ort

    def __repr__(self):
        return f"Adresse('{self.strasse}', '{self.plz}', '{self.ort}')"


class Projekt:
    def __init__(self, name, beschreibung):
        self.name = name
        self.beschreibung = beschreibung

    def __repr__(self):
        return f"Projekt('{self.name}', '{self.beschreibung}')"


class Mitarbeiter:
    def __init__(self, vorname, nachname, personalnummer, projekt=None):
        self.vorname = vorname
        self.nachname = nachname
        self.personalnummer = personalnummer
        self.adressen = []
        self.projekt = projekt

    def add_adresse(self, adresse):
        if isinstance(adresse, Adresse):
            self.adressen.append(adresse)

    def remove_adresse(self, adresse):
        if adresse in self.adressen:
            self.adressen.remove(adresse)

    def set_projekt(self, projekt):
        if isinstance(projekt, Projekt):
            self.projekt = projekt

    def __repr__(self):
        return f"Mitarbeiter('{self.vorname}', '{self.nachname}', {self.personalnummer})"


class Firma:
    def __init__(self, name, adresse=None, bekannter_mitarbeiter=None):
        self.name = name
        self.adresse = adresse
        self.mitarbeiter = []
        self.bekannter_mitarbeiter = bekannter_mitarbeiter

    def add_mitarbeiter(self, mitarbeiter):
        if isinstance(mitarbeiter, Mitarbeiter):
            self.mitarbeiter.append(mitarbeiter)

    def remove_mitarbeiter(self, mitarbeiter):
        if mitarbeiter in self.mitarbeiter:
            self.mitarbeiter.remove(mitarbeiter)

    def set_adresse(self, adresse):
        if isinstance(adresse, Adresse):
            self.adresse = adresse

    def set_bekannter_mitarbeiter(self, mitarbeiter):
        if isinstance(mitarbeiter, Mitarbeiter):
            self.bekannter_mitarbeiter = mitarbeiter

    def __repr__(self):
        return f"Firma('{self.name}')"
```

```python
adresse_firma = Adresse("Leopoldstraße 10", "80802", "München")
adresse_1 = Adresse("Rosenstraße 5", "80331", "München")
adresse_2 = Adresse("Isarweg 7", "81541", "München")
```

```python
projekt_1 = Projekt("KI-Projekt", "Python-Projekt mit OOP")
```

```python
mitarbeiter_1 = Mitarbeiter("Max", "Müller", 1001)
mitarbeiter_2 = Mitarbeiter("Anna", "Schmidt", 1002)
```

```python
mitarbeiter_1.add_adresse(adresse_1)
mitarbeiter_1.add_adresse(adresse_2)
```

```python
mitarbeiter_1.remove_adresse(adresse_2)
```

```python
mitarbeiter_1.set_projekt(projekt_1)
```

```python
firma = Firma("Coding GmbH")
firma.set_adresse(adresse_firma)
```

```python
firma.add_mitarbeiter(mitarbeiter_1)
firma.add_mitarbeiter(mitarbeiter_2)
```

```python
firma.set_bekannter_mitarbeiter(mitarbeiter_1)
```

```python
print("Firma:")
print(firma)
```

```python
print("Adresse der Firma:")
print(firma.adresse)
```

```python
print("Mitarbeiter der Firma:")
for m in firma.mitarbeiter:
    print(m)
```

```python
print("Bekannter Mitarbeiter:")
print(firma.bekannter_mitarbeiter)
```

```python
print("Adressen von Mitarbeiter 1:")
for adresse in mitarbeiter_1.adressen:
    print(adresse)
```
