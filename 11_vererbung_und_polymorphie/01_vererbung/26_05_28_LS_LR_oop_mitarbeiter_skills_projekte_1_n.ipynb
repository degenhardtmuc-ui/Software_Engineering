```python
# Klasse A kennt/hat 1-N von Klasse B
# Dann hat klasse A eine Listen-Referenz auf B

class Person: 
    
    def __init__(self, name, alter):
        self.name = name
        self.alter = alter
        self.addressen = []
        
    
    def add_addresse(self, addresse):
        if isinstance(addresse, Addresse):
            self.addressen.append(addresse)
            
        
    def __repr__(self):
        return f"Person('{self.name}', {self.alter}, addressen = {self.addressen})"
```
