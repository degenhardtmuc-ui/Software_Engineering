```python
class Addresse:

    def __init__(self, strasse, nr, plz, stadt):
        self.strasse = strasse
        self.nr = nr
        self.plz = plz
        self.stadt = stadt

    def __repr__(self):
        return f"Addresse('{self.strasse}', '{self.nr}', {self.plz}, '{self.stadt}')"


class Person:

    def __init__(self, name, alter, addresse=None):
        self.name = name
        self.alter = alter
        self.addresse = addresse

    def __repr__(self):
        return f"Person('{self.name}', {self.alter}, addresse = {self.addresse})"


# Adresse erzeugen
home = Addresse("Alte Alle", "5b", 81245, "München")

# Person erzeugen und mit Adresse verknüpfen
barbara = Person("Barbara", 27, home)

# Ausgabe
print(barbara)

# Stadt ändern
home.stadt = "Hannover"

# Neue Ausgabe
print(home)

# Person erneut ausgeben
print(barbara)

# Straße über die Person ändern
barbara.addresse.strasse = "Sesam Strasse"

# Adresse erneut ausgeben
print(home)
```
