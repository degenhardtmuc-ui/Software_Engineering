```python
from abc import ABC, abstractmethod #Wir holen uns aus Python Werkzeuge, um eine abstrakte Klasse / ein Interface-ähnliches Verhalten zu bauen
# abc/ABC = Abstract Base Classes
# Auf Deutsch:
# abstrakte Basisklassen
# Also Klassen, die als Vorlage dienen

class Animal(ABC): #Animal ist die allgemeine Klasse.Sie funktioniert hier wie ein Interface.

    @abstractmethod
    def make_sound(self): #Jedes Tier muss eine Methode make_sound() haben
        pass


class Dog(Animal):

    def make_sound(self): #Dog erbt von Animal
        print("Wuff")


class Cat(Animal):

    def make_sound(self): #Cat erbt auch von Animal
        print("Miau")


animals = [Dog(), Cat(), Dog(), Cat()]

for animal in animals:
    animal.make_sound()
```

```python
from abc import ABC, abstractmethod


class Animal(ABC):

    @abstractmethod
    def make_sound(self):
        pass


class Dog(Animal):

    def make_sound(self):
        print("Wuff")


class Cat(Animal):

    def make_sound(self):
        print("Miau")


dog = Dog()
cat = Cat()

dog.make_sound()
cat.make_sound()
```
