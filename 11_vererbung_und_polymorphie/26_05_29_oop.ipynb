```python
class Mitarbeiter(Person): # Unterklasse, Kindklasse
    
    def __init__(self, name, alter, gehalt):
        super().__init__(name, alter) # => rufe ich ___init__ von der Klasse Person
        self.gehalt = gehalt
        
    def get_description(self):
        info = super().get_description()
        result = info + " " + f"und mein Gehalt ist {self.gehalt}" 
        return result
```

```python
from dataclasses import dataclass

@dataclass
class Person:  # Oberklasse, Wurzelklasse, Vaterklasse
    
    name: str
    alter: int

    def __repr__(self):
        return f"Person('{self.name}', {self.alter})"
    
    def get_description(self):
        return f"Ich heiße {self.name} und bin {self.alter} Jahre alt"

class Student(Person):  # Unterklasse, Kindklasse, Tochterklasse
    
    def __init__(self, name, alter, studienfach):
        super().__init__(name, alter)  # ruft den Konstruktor der Oberklasse auf
        self.studienfach = studienfach
        
    def __repr__(self):
        return f"Student('{self.name}', {self.alter}, '{self.studienfach}')"
    
    def get_description(self):
        super_description = super().get_description()  # ruft die Methode der Oberklasse auf
        return f"{super_description} und studiere {self.studienfach}"
```
