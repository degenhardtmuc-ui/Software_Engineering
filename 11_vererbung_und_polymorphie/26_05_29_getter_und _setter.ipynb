```python
class Person:
    def __init__(self, name, alter):
        self.name = name
        self._alter = alter  # Das eigentliche, "versteckte" Attribut

    # Getter: Ruft den Wert ab
    @property
    def alter(self):
        return self._alter

    # Setter: Setzt den Wert mit Validierungslogik
    @alter.setter
    def alter(self, neuer_wert):
        if neuer_wert < 0:
            raise ValueError("Das Alter kann nicht negativ sein!")
        self._alter = neuer_wert

# Verwendung (Beachten Sie: Sie rufen es auf wie ein normales Attribut)
p = Person("Anna", 25)
p.alter = 30  # Ruft automatisch den Setter auf
print(p.alter)  # Ruft automatisch den Getter auf
```

```python
class Person:
    def init(self):
        self._age = 0
    # getter
    def get_age(self):
        return self._age
    # setter
    def set_age(self, value):
        if value < 0:
            raise ValueError("Age cannot be negative")
        self._age = value
p = Person()
p.set_age(25)
print(p.get_age())
```
