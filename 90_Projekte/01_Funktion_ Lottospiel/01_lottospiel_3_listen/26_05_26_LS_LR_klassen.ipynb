```python
# Klasse Hund
# Klasse Mitarbeiter
# Klasse Student
# Klasse Buerger


class Hund:
    def __init__(self, name=None, alter=0, rasse=None):
        self.name = name
        self.alter = alter
        self.rasse = rasse


class Mitarbeiter:
    def __init__(self, name=None, alter=0, mitarbeiternummer=None, abteilung=None):
        self.name = name
        self.alter = alter
        self.mitarbeiternummer = mitarbeiternummer
        self.abteilung = abteilung


class Student:
    def __init__(self, name=None, alter=0, matrikelnummer=None, studiengang=None):
        self.name = name
        self.alter = alter
        self.matrikelnummer = matrikelnummer
        self.studiengang = studiengang


class Buerger:
    def __init__(self, name=None, alter=0, ausweisnummer=None, stadt=None):
        self.name = name
        self.alter = alter
        self.ausweisnummer = ausweisnummer
        self.stadt = stadt


# Objekte erzeugen

hund1 = Hund("Bello", 5, "Labrador")
mitarbeiter1 = Mitarbeiter("Daniel", 42, 1001, "IT")
student1 = Student("Anna", 22, 123456, "Informatik")
buerger1 = Buerger("Leonid", 40, "MUC123456", "Muenchen")


# einzelne Attribute testen

print(hund1.name)
print(hund1.rasse)

print(mitarbeiter1.name)
print(mitarbeiter1.mitarbeiternummer)

print(student1.name)
print(student1.studiengang)

print(buerger1.name)
print(buerger1.ausweisnummer)
```
