```python
spieler_zahlen = [3, 7, 12, 25, 33, 49]

lotto_reihen = [
    [1, 7, 12, 18, 33, 45],
    [2, 4, 8, 16, 23, 42],
    [3, 5, 12, 25, 38, 49]
]

for lotto_zahlen in lotto_reihen:
    treffer = []

    for zahl in spieler_zahlen:
        if zahl in lotto_zahlen:
            treffer.append(zahl)
```

```python
print("Lottozahlen:", lotto_zahlen)
print("Treffer:", treffer)
print("Anzahl Treffer:", len(treffer))
```
