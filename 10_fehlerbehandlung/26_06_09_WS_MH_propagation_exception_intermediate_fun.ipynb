```python
from typing import Callable


def intermediate_fun_v0(error_type: Callable):
    try:
        raise_and_handle_error_v0(error_type)
    except:
        raise


def raise_and_handle_error_v0(error_type: Callable):
    try:
        raise error_type(f"Raising {error_type.__name__}")
    except LookupError:
        pass



# TESTBEREICH


print("Test 1: KeyError")
intermediate_fun_v0(KeyError)
print("KeyError wurde in raise_and_handle_error_v0 gefangen.")
print()


print("Test 2: IndexError")
intermediate_fun_v0(IndexError)
print("IndexError wurde in raise_and_handle_error_v0 gefangen.")
print()


print("Test 3: ValueError")
try:
    intermediate_fun_v0(ValueError)
except ValueError as error:
    print("ValueError wurde NICHT innen gefangen.")
    print("Der Fehler wurde nach außen weitergegeben.")
    print("Fehlermeldung:", error)
print()


print("Test 4: TypeError")
try:
    intermediate_fun_v0(TypeError)
except TypeError as error:
    print("TypeError wurde NICHT innen gefangen.")
    print("Der Fehler wurde nach außen weitergegeben.")
    print("Fehlermeldung:", error)
```
