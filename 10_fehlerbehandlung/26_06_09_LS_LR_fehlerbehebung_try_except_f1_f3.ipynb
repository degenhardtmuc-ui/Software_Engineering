```python
def f1():
    
    print("in f1() call f2()")

    try:
        # f1() ruft f2() auf.
        # Wenn in f2() oder tiefer ein Fehler passiert und NICHT abgefangen wird,
        # dann wuerde der Fehler hier im except von f1() landen.
        f2()

        # Diese Zeilen laufen nur, wenn f2() am Ende normal zurueckkommt.
        print("f2() erfolgreich ausgefuehrt")
        print("Also weiter machen mit f1()")

    except Exception as e:
        # Dieser Block wird in diesem Beispiel NICHT ausgefuehrt,
        # weil f3() den Fehler schon vorher abfaengt.
        print("Exception wurde in f1() behandelt")
        print("Fehlermeldung:", e)

    print("Fertig mit f1()")


def f2():
    
    print("in f2() call f3()")

    # f2() ruft f3() auf.
    f3()

    # Diese Zeilen werden ausgefuehrt, weil f3() den Fehler abfaengt
    # und danach normal zu f2() zurueckkommt.
    print("weiter machen mit f2()")
    print("Fertig mit f2()")


def f3():
    
    try:
        print("in f3() call f4()")

        # f3() ruft f4() auf.
        # In f4() wird spaeter fN() aufgerufen.
        # Dort entsteht der Fehler.
        f4()

        # Diese Zeilen werden NICHT ausgefuehrt,
        # weil f4() wegen des Fehlers nicht normal zurueckkommt.
        print("weiter machen mit f3()")
        print("Fertig mit f3() im try")

    except Exception as e:
        # Hier landet der Fehler aus fN().
        # f3() ist naeher am Fehler als f1(), deshalb faengt f3() ihn zuerst ab.
        print("Exception wurde in f3() behandelt")
        print("Fehlermeldung:", e)

    # Nach dem except geht es hier weiter.
    print("Fertig mit f3()")


def f4():
  
    print("in f4() call fN()")

    # fN() loest absichtlich einen Fehler aus.
    fN()

    # Diese Zeilen werden NICHT ausgefuehrt,
    # weil fN() vorher mit raise einen Fehler wirft.
    print("fN() erfolgreich ausgefuehrt")
    print("Also weiter machen mit f4()")
    print("Fertig mit f4()")


def fN():
    
    raise ValueError("Raising ValueError")

    # Diese Zeile wird NIE ausgefuehrt,
    # weil nach raise der normale Ablauf sofort stoppt.
    print("code nach ValueError in fN()")


# Programm starten:
# Erst durch diesen Aufruf beginnt die Kette wirklich zu laufen.
f1()
```

```python
def f1():
    try:
        print("in f1() call f2()")
        f2()
        print("f2() erfolgreich ausgeführt")
        print("Fertig mit f1()")
    except Exception:
        print("Exception wurde in f1() behandelt")


def f2():
    print("in f2() call f3()")
    f3()
    print("weiter machen mit f2()")
    print("Fertig mit f2()")


def f3():
    try:
        print("in f3() call f4()")
        f4()
        print("weiter machen mit f3()")
        print("Fertig mit f3()")
    except Exception:
        print("Exception wurde in f3() behandelt")
    else:
        print("kein Fehler in f3() passiert") # läuft nur weiter, wenn kein Fehler passiert
    finally:
        print("will always run") # wird immer ausgeführt, egal ob Fehler oder kein Fehler

def f4():
    print("in f4() call fN()")
    fN()
    print("weiter machen mit f4()")


def fN():
    raise ValueError("Raising ValueError")


f1()
```
