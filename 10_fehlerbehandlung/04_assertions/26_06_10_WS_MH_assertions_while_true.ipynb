```python
valid_mail_addresses = [
    "anna@example.com",
    "max@example.com",
    "daniel@example.com",
]

while True:
    mail_adresse = input("Bitte E-Mail-Adresse eingeben: ")

    if mail_adresse in valid_mail_addresses:
        print(f"Gültige E-Mail-Adresse: {mail_adresse}")
        break

    print("Bitte gib eine gültige E-Mail-Adresse ein.")
```
